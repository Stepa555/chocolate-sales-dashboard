```python
import pandas as pd
```

```python
df = pd.read_csv('Chocolate_Sales.csv')
```

```python
# Посмотрим первые 5 строк
print("Первые 5 строк:")
print(df.head())

# Посмотрим информацию о данных
print("\nИнформация о данных:")
print(df.info())

# Посмотрим описательную статистику
print("\nОписательная статистика:")
print(df.describe())
```

```python
# Сколько пропусков в каждом столбце

print('\nПропуски в данных')
print(df.isnull().sum())
```

```python
# Очищаем колонку Amount
# 1. Убираем символы $ и пробелы
df['Amount'] = df['Amount'].astype(str).str.replace('$', '').str.strip()

# 2. Заменяем пустые строки на NaN
df['Amount'] = df['Amount'].replace('', pd.NA)

# 3. Преобразуем в числа (если не получается — ставим NaN)
df['Amount'] = pd.to_numeric(df['Amount'], errors='coerce')

# 4. Удаляем строки, где Amount стал NaN (или можно заполнить)
df = df.dropna(subset=['Amount'])

# Проверяем, что получилось
print(f"Осталось строк после очистки: {len(df)}")
print(df[['Amount']].head(10))
```

```python
# Ключевые метрики

# 2. Проверяем колонки
print("Доступные колонки:", list(df.columns))

# 3. Выбираем колонку для анализа
amount_col = input("Введите название колонки с суммой: ")

# 4. Очищаем данные
df[amount_col] = pd.to_numeric(df[amount_col], errors='coerce')
df = df.dropna(subset=[amount_col])

# 5. Считаем метрики
total = df[amount_col].sum()
average = df[amount_col].mean()
orders = len(df)

# 6. Выводим с округлением
print("\n" + "="*40)
print(f"Общая выручка: {total/1_000_000:.2f} млн руб.")
print(f"Средний чек: {average:,.2f} руб.")
print(f"Количество заказов: {orders:,}")
```

## Выводы после первичного анализа:
- Всего строк: 200000
- Количество столбцов: 11
- Есть пропуски в колонках: Order_Date, Discount_Pct, Price_per_Box,    Marketing_Spend
- Основные метрики: общая выручка ~ 102,572.4 тыс. руб, средний чек ~ 512.86 руб.
